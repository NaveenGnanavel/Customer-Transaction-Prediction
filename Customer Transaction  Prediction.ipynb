```python
import joblib
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix,
    RocCurveDisplay
)
```

```python
X = df.drop(columns=["ID_code", "target"],errors="ignore")
y = df["target"]
X = X.apply(pd.to_numeric, errors="coerce")
X = X.replace([np.inf, -np.inf],np.nan)
X = X.dropna(axis=1, how="all")
print("Input Features:", X.shape)
print("Target Shape:", y.shape)
print("Missing Values:", X.isnull().sum().sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
print("\nTraining Target Distribution:")
print(y_train.value_counts())
```

```python
models = {
    "Logistic Regression": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ))
    ]),
    "Random Forest": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", RandomForestClassifier(
            n_estimators=100,
            max_depth=15,
            min_samples_leaf=3,
            class_weight="balanced_subsample",
            random_state=42,
            n_jobs=-1
        ))
    ]),
    "Extra Trees": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", ExtraTreesClassifier(
            n_estimators=100,
            max_depth=15,
            min_samples_leaf=3,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ))
    ])
}
print("Machine Learning Models Created Successfully!")
```

```python
results = []
trained_models = {}
for name, model in models.items():
    print("\nTraining Model:", name)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob),
        "PR-AUC": average_precision_score(y_test, y_prob)
    })
    trained_models[name] = model
print("\nModel Training Completed!")
```

```python
results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by="ROC-AUC",ascending=False).reset_index(drop=True)
display(results_df.round(4))
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=results_df,x="Model",y="ROC-AUC")
plt.title("Machine Learning Model Comparison")
plt.xlabel("Models")
plt.ylabel("ROC-AUC Score")
plt.xticks(rotation=15)
plt.show()
```

```python
best_model_name = results_df.loc[0, "Model"]
best_model = trained_models[best_model_name]
print("Best Model:", best_model_name)
print("Best ROC-AUC Score:",round(results_df.loc[0, "ROC-AUC"], 4))
```

```python
y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]
print("Classification Report:")
print(classification_report(y_test,y_pred,zero_division=0))
print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
print("PR-AUC:", average_precision_score(y_test, y_prob))
```

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm,annot=True,fmt="d",cmap="Blues")
plt.title("Best Model - Confusion Matrix")
plt.xlabel("Predicted Values")
plt.ylabel("Actual Values")
plt.show()
```

```python
RocCurveDisplay.from_predictions(y_test,y_prob)
plt.title("ROC Curve - Best Model")
plt.show()
```

```python
best_classifier = best_model.named_steps["classifier"]
if hasattr(best_classifier, "feature_importances_"):
    importance_df = pd.DataFrame({"Feature": X.columns,"Importance": best_classifier.feature_importances_})
    importance_df = importance_df.sort_values(by="Importance",ascending=False)
    print("Top 20 Important Features:")
    display(importance_df.head(20))
    plt.figure(figsize=(10, 7))
    sns.barplot(data=importance_df.head(20),x="Importance",y="Feature")
    plt.title("Top 20 Feature Importance")
    plt.show()
else:
    print("Selected model does not provide direct feature importance.")
```

```python
joblib.dump(best_model,"customer_transaction_model.pkl")
print("Model saved successfully!")
```

```python
prediction_results = pd.DataFrame({"Actual_Target": y_test.values,"Predicted_Target": y_pred,"Transaction_Probability": y_prob})
display(prediction_results.head(20))
prediction_results.to_csv("customer_transaction_predictions.csv",index=False)
print("Test predictions saved successfully!")
```

```python
future_probabilities = best_model.predict_proba(X_test)[:, 1]
future_predictions = (future_probabilities >= 0.5).astype(int)
future_results = pd.DataFrame({
    "ID_code": df.loc[X_test.index, "ID_code"].values,
    "Transaction_Probability": future_probabilities,
    "Predicted_Target": future_predictions,
    "Actual_Target": y_test.values
})
display(future_results.head(20))
future_results.to_csv("future_customer_predictions.csv",index=False)
print("Customer Transaction Predictions Completed!")
```

## MODEL COMPARISON REPORT

### 1. Objective
The objective of this comparison is to evaluate multiple machine learning classification algorithms and identify the most suitable model for predicting future customer transactions.

### 2. Models Implemented
The following classification algorithms were implemented:

- **Logistic Regression:** Used as a baseline classification model to identify relationships between input features and transaction outcomes.
- **Random Forest Classifier:** An ensemble learning algorithm that combines multiple decision trees to improve prediction performance and reduce overfitting.
- **Extra Trees Classifier:** An ensemble algorithm that uses randomized decision trees to improve model generalization and classification performance.

### 3. Evaluation Metrics
The models were evaluated using the following performance metrics:

- **Accuracy:** Measures the overall percentage of correct predictions.
- **Precision:** Measures how many predicted transaction customers actually belong to the positive class.
- **Recall:** Measures how many actual transaction customers are correctly identified.
- **F1 Score:** Provides a balance between Precision and Recall.
- **ROC-AUC:** Measures the model's ability to distinguish between transaction and non-transaction customers.
- **PR-AUC:** Evaluates positive-class prediction performance, particularly useful when the target classes are imbalanced.

### 4. Model Performance Comparison
A comparative analysis was performed using the evaluation metrics. The results were organized into a model comparison table to identify the best-performing algorithm.

The model with the highest ROC-AUC score was selected as the best-performing model.

### 5. Best Model Recommendation for Production
The model achieving the highest ROC-AUC score is recommended for further production evaluation because it provides better discrimination between customers who are likely and unlikely to perform transactions.

Before actual banking deployment, the selected model should undergo additional validation, probability calibration, threshold optimization, and fairness testing.

### 6. Conclusion
The comparison of Logistic Regression, Random Forest, and Extra Trees helps identify the most suitable classification algorithm for customer transaction prediction. The final model selection is based on its overall evaluation performance, with particular importance given to ROC-AUC and PR-AUC.

## REPORT ON CHALLENGES FACED

During the Banking Customer Transaction Prediction project, several challenges were encountered while analyzing the dataset and developing the predictive model.

### 1. Anonymized Feature Names
**Challenge:** The dataset contained 200 anonymized numerical features without meaningful business descriptions.

**Technique Used:** Statistical analysis, descriptive statistics, feature distribution analysis, and correlation analysis.

**Reason:** These techniques helped understand numerical patterns and relationships without requiring feature descriptions.

### 2. Target Class Imbalance
**Challenge:** The target variable contained two classes (0 and 1), with a potential imbalance between transaction and non-transaction customers.

**Technique Used:** Target distribution analysis, stratified train-test splitting, and class weighting.

**Reason:** These techniques helped preserve class proportions and improve the model's ability to identify minority-class customers.

### 3. Missing Values
**Challenge:** Missing values can affect machine learning model training and prediction.

**Technique Used:** SimpleImputer with median imputation.

**Reason:** Median imputation provides a practical way to handle missing numerical values while reducing the influence of extreme values.

### 4. Feature Scaling
**Challenge:** Numerical features may have different value ranges, which can affect certain machine learning algorithms.

**Technique Used:** StandardScaler.

**Reason:** Feature scaling standardizes numerical values and improves the training process of Logistic Regression.

### 5. Model Selection
**Challenge:** Selecting the most suitable algorithm for customer transaction prediction.

**Technique Used:** Compared Logistic Regression, Random Forest, and Extra Trees using Accuracy, Precision, Recall, F1 Score, ROC-AUC, and PR-AUC.

**Reason:** Comparing multiple evaluation metrics provides a more reliable understanding of model performance than relying only on accuracy.

### 6. Prediction Dataset Availability
**Challenge:** The separate test dataset was not available in the initial working directory.

**Technique Used:** Used the validation dataset generated through train-test splitting for model evaluation and prediction.

**Reason:** This allowed model predictions to be tested against known actual target values. A separate unseen dataset is still required for genuine future customer prediction.

### 7. Feature Interpretation
**Challenge:** The anonymized features could not be directly associated with specific customer or banking activities.

**Technique Used:** Feature importance analysis.

**Reason:** This helped identify influential numerical variables without assigning unsupported business meanings to them.

### Conclusion
The challenges were addressed using suitable data preprocessing, statistical analysis, feature scaling, class balancing, and model evaluation techniques. These methods helped develop a structured machine learning pipeline for predicting future customer transactions.
